```python
from mpi4py import MPI
from dolfinx import mesh
import numpy
from dolfinx import fem
```

```python
# Create the mesh
domain = mesh.create_unit_square(MPI.COMM_WORLD, 8, 8, mesh.CellType.quadrilateral)
```

```python
# Define the finite element function space V.
# We need to define where the space is defined on, the element family
# the space is based on, and the degree of the element.
V = fem.functionspace(domain, ("Lagrange", 1))
```

```python
# Here we are setting up the Dirchlet boundary conditions
# on the domain edge. Notice that x[0] = x and x[1] = y.
uD = fem.Function(V)
uD.interpolate(lambda x: 1 + x[0] ** 2 + 2 * x[1] ** 2)
```

```python
tdim = domain.topology.dim # Dimensions of the elements
fdim = tdim - 1 # Dimensions of boundaries
domain.topology.create_connectivity(fdim, tdim) # Map between elements and their facets
boundary_facets = mesh.exterior_facet_indices(domain.topology) # Get the boundaries
```

```python
boundary_dofs = fem.locate_dofs_topological(V, fdim, boundary_facets)
bc = fem.dirichletbc(uD, boundary_dofs)
```

```python
# Now we define the trial and test functions.
import ufl
u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)
```

```python
# Define the source term
from dolfinx import default_scalar_type
f = fem.Constant(domain, default_scalar_type(-6))
```

```python
# Create weak formulation
a = ufl.dot(ufl.grad(u), ufl.grad(v)) * ufl.dx # stiffness matrix
L = f * v * ufl.dx # force term
```

```python
# Forming and solving the linear system
from dolfinx.fem.petsc import LinearProblem
problem = LinearProblem(
    a, 
    L,
    bcs=[bc],
    petsc_options={"ksp_type": "preonly", "pc_type": "lu"},
    petsc_options_prefix="Poisson",
)
uh = problem.solve()
```
